```python
from pathlib import Path

RAW = Path("data/luad_raw")

files = list(RAW.glob("LUAD*"))

print(f"Found {len(files)} LUAD files:\n")
for f in files:
    print(f.name)
```

```python
import pandas as pd

rna_tumor_file = next(RAW.glob("*RNAseq*Tumor.txt"))
rna_normal_file = next(RAW.glob("*RNAseq*Normal.txt"))

rna_tumor = pd.read_csv(rna_tumor_file, sep="\t")
rna_normal = pd.read_csv(rna_normal_file, sep="\t")

print("RNA Tumor shape:", rna_tumor.shape)
print("RNA Normal shape:", rna_normal.shape)

print("\nRNA Tumor first 5 rows:")
display(rna_tumor.head())

print("\nRNA Normal first 5 rows:")
display(rna_normal.head())
```

```python
# Load proteomics and mutation files

prot_tumor_file = next(RAW.glob("*proteomics*Tumor.txt"))
prot_normal_file = next(RAW.glob("*proteomics*Normal.txt"))
mut_file = next(RAW.glob("*somatic_mutation*gene_level*"))

prot_tumor = pd.read_csv(prot_tumor_file, sep="\t")
prot_normal = pd.read_csv(prot_normal_file, sep="\t")
mutation = pd.read_csv(mut_file, sep="\t")

print("Protein Tumor shape:", prot_tumor.shape)
print("Protein Normal shape:", prot_normal.shape)
print("Mutation shape:", mutation.shape)

print("\nProtein Tumor:")
display(prot_tumor.head())

print("\nProtein Normal:")
display(prot_normal.head())

print("\nMutation:")
display(mutation.head())
```

```python
print("Mutation columns:")
print(mutation.columns[:10].tolist())

print("\nMutation first 10 rows:")
display(mutation.head(10))
```

```python
# Remove Ensembl version numbers so IDs match cleanly

rna_tumor["ensembl"] = rna_tumor["idx"].str.split(".").str[0]
rna_normal["ensembl"] = rna_normal["idx"].str.split(".").str[0]

prot_tumor["ensembl"] = prot_tumor["idx"].str.split(".").str[0]
prot_normal["ensembl"] = prot_normal["idx"].str.split(".").str[0]

mutation["ensembl"] = mutation["idx"].str.split(".").str[0]

# Check overlap across the 3 omics layers
rna_genes = set(rna_tumor["ensembl"])
prot_genes = set(prot_tumor["ensembl"])
mut_genes = set(mutation["ensembl"])

common = rna_genes & prot_genes & mut_genes

print("RNA genes:", len(rna_genes))
print("Protein genes:", len(prot_genes))
print("Mutation genes:", len(mut_genes))
print("Common genes across all 3 layers:", len(common))
```

```python
# Check sample overlap between Tumor and Normal

rna_tumor_samples = set(rna_tumor.columns) - {"idx", "ensembl"}
rna_normal_samples = set(rna_normal.columns) - {"idx", "ensembl"}

prot_tumor_samples = set(prot_tumor.columns) - {"idx", "ensembl"}
prot_normal_samples = set(prot_normal.columns) - {"idx", "ensembl"}

print("RNA Tumor samples:", len(rna_tumor_samples))
print("RNA Normal samples:", len(rna_normal_samples))
print("RNA matched Tumor/Normal samples:", len(rna_tumor_samples & rna_normal_samples))

print()

print("Protein Tumor samples:", len(prot_tumor_samples))
print("Protein Normal samples:", len(prot_normal_samples))
print("Protein matched Tumor/Normal samples:", len(prot_tumor_samples & prot_normal_samples))
```

```python
import numpy as np
from scipy import stats

# -------------------------
# RNA: paired Tumor vs Normal
# -------------------------

rna_pairs = sorted(rna_tumor_samples & rna_normal_samples)

# Collapse any duplicate Ensembl IDs after removing version numbers
rna_t = (
    rna_tumor[["ensembl"] + rna_pairs]
    .groupby("ensembl")
    .mean()
)

rna_n = (
    rna_normal[["ensembl"] + rna_pairs]
    .groupby("ensembl")
    .mean()
)

# Keep genes present in both
rna_common = rna_t.index.intersection(rna_n.index)
rna_t = rna_t.loc[rna_common]
rna_n = rna_n.loc[rna_common]

# Because the source values are already log2 transformed,
# Tumor - Normal = log2 fold change
rna_log2fc = (rna_t - rna_n).mean(axis=1)

rna_pval = stats.ttest_rel(
    rna_t.to_numpy(),
    rna_n.to_numpy(),
    axis=1,
    nan_policy="omit"
).pvalue

rna_summary = pd.DataFrame({
    "ensembl": rna_common,
    "log2fc": rna_log2fc.values,
    "pval": rna_pval
})

print("RNA summary shape:", rna_summary.shape)
display(rna_summary.head())
```

```python
# -------------------------
# Protein: paired Tumor vs Normal
# -------------------------

prot_pairs = sorted(prot_tumor_samples & prot_normal_samples)

# Collapse duplicate Ensembl IDs if any
prot_t = (
    prot_tumor[["ensembl"] + prot_pairs]
    .groupby("ensembl")
    .mean()
)

prot_n = (
    prot_normal[["ensembl"] + prot_pairs]
    .groupby("ensembl")
    .mean()
)

# Keep genes present in both datasets
prot_common = prot_t.index.intersection(prot_n.index)
prot_t = prot_t.loc[prot_common]
prot_n = prot_n.loc[prot_common]

# Paired differences
prot_diff = prot_t - prot_n

# Mean paired difference = log2 fold change
prot_log2fc = prot_diff.mean(axis=1, skipna=True)

# Number of valid matched pairs for each gene
valid_pairs = prot_diff.notna().sum(axis=1)

# Paired t-test, ignoring missing pairs
prot_pval = stats.ttest_rel(
    prot_t.to_numpy(),
    prot_n.to_numpy(),
    axis=1,
    nan_policy="omit"
).pvalue

prot_summary = pd.DataFrame({
    "ensembl": prot_common,
    "log2fc": prot_log2fc.values,
    "pval": prot_pval,
    "n_pairs": valid_pairs.values
})

print("Protein summary shape:", prot_summary.shape)

print("\nValid pair counts:")
print(prot_summary["n_pairs"].describe())

display(prot_summary.head())
```

```python
# -------------------------
# Genomics: mutation frequency
# -------------------------

mut_sample_cols = [
    c for c in mutation.columns
    if c not in ["idx", "ensembl"]
]

# Collapse duplicate Ensembl IDs if any
mut_matrix = (
    mutation[["ensembl"] + mut_sample_cols]
    .groupby("ensembl")
    .mean()
)

# Fraction of tumors carrying a mutation in each gene
mutation_freq = mut_matrix.mean(axis=1, skipna=True)

mut_summary = pd.DataFrame({
    "ensembl": mut_matrix.index,
    "mutation_freq": mutation_freq.values
})

print("Mutation summary shape:", mut_summary.shape)

print("\nMutation frequency summary:")
print(mut_summary["mutation_freq"].describe())

print("\nTop 10 most frequently mutated genes:")
display(
    mut_summary.sort_values("mutation_freq", ascending=False).head(10)
)
```

```python
# Rename columns before merging so nothing collides

rna_final = rna_summary.rename(columns={
    "log2fc": "rna_log2fc",
    "pval": "rna_pval"
})

prot_final = prot_summary.rename(columns={
    "log2fc": "prot_log2fc",
    "pval": "prot_pval"
})

# Merge RNA + protein + mutation
merged = (
    rna_final
    .merge(prot_final, on="ensembl", how="inner")
    .merge(mut_summary, on="ensembl", how="inner")
)

print("Genes surviving all 3 layers:", merged.shape[0])

display(merged.head())
```

```python
import mygene
import pandas as pd

mg = mygene.MyGeneInfo()

results = mg.querymany(
    merged["ensembl"].tolist(),
    scopes="ensembl.gene",
    fields="symbol",
    species="human"
)

mapping = pd.DataFrame([
    {
        "ensembl": x.get("query"),
        "gene": x.get("symbol")
    }
    for x in results
    if x.get("symbol") is not None
])

mapping = mapping.drop_duplicates(subset="ensembl")

print("Ensembl IDs submitted:", len(merged))
print("Successfully mapped:", len(mapping))

display(mapping.head(10))
```

```python
merged_genes = merged.merge(mapping, on="ensembl", how="inner")

print("Genes after symbol mapping:", len(merged_genes))
print("Duplicate gene symbols:", merged_genes["gene"].duplicated().sum())

display(
    merged_genes[
        [
            "gene",
            "ensembl",
            "rna_log2fc",
            "rna_pval",
            "prot_log2fc",
            "prot_pval",
            "mutation_freq",
            "n_pairs"
        ]
    ].head(10)
)
```

```python
import numpy as np

df = merged_genes.copy()

# 1. RNA vs protein direction agreement
df["concordant"] = (
    np.sign(df["rna_log2fc"]) ==
    np.sign(df["prot_log2fc"])
)

# 2. Evidence magnitude for each layer
df["transcriptomic"] = df["rna_log2fc"].abs()
df["proteomic"] = df["prot_log2fc"].abs()
df["genomic"] = df["mutation_freq"]

# 3. Rank-normalize each layer to 0–1
df["tx_rank"] = df["transcriptomic"].rank(pct=True)
df["prot_rank"] = df["proteomic"].rank(pct=True)
df["gen_rank"] = df["genomic"].rank(pct=True)

# 4. Equal-weight multi-evidence score
df["score"] = (
    df["tx_rank"] +
    df["prot_rank"] +
    df["gen_rank"]
) / 3

# 5. Rank highest to lowest
ranked_luad = df.sort_values("score", ascending=False).reset_index(drop=True)

print("Genes scored:", len(ranked_luad))
print("Concordant genes:", ranked_luad["concordant"].sum())
print("Discordant genes:", (~ranked_luad["concordant"]).sum())

display(
    ranked_luad[
        [
            "gene",
            "rna_log2fc",
            "rna_pval",
            "prot_log2fc",
            "prot_pval",
            "mutation_freq",
            "n_pairs",
            "concordant",
            "score"
        ]
    ].head(15)
)
```

```python
# Add overall rank
ranked_luad["rank"] = ranked_luad.index + 1

# Check established LUAD-related genes
known_luad = [
    "TP53", "KRAS", "EGFR", "STK11",
    "KEAP1", "BRAF", "MET", "ERBB2",
    "ALK", "NF1"
]

known_hits = ranked_luad[
    ranked_luad["gene"].isin(known_luad)
][
    [
        "rank",
        "gene",
        "rna_log2fc",
        "prot_log2fc",
        "mutation_freq",
        "n_pairs",
        "concordant",
        "score"
    ]
]

display(known_hits.sort_values("rank"))
```

```python
print("Top 15 protein coverage:")
display(
    ranked_luad[
        ["rank", "gene", "n_pairs", "score"]
    ].head(15)
)
```

```python
# QC filter: require protein measurements in at least 50 matched pairs
df_qc = merged_genes[merged_genes["n_pairs"] >= 50].copy()

# Concordance
df_qc["concordant"] = (
    np.sign(df_qc["rna_log2fc"]) ==
    np.sign(df_qc["prot_log2fc"])
)

# Evidence magnitudes
df_qc["transcriptomic"] = df_qc["rna_log2fc"].abs()
df_qc["proteomic"] = df_qc["prot_log2fc"].abs()
df_qc["genomic"] = df_qc["mutation_freq"]

# Normalize each layer
df_qc["tx_rank"] = df_qc["transcriptomic"].rank(pct=True)
df_qc["prot_rank"] = df_qc["proteomic"].rank(pct=True)
df_qc["gen_rank"] = df_qc["genomic"].rank(pct=True)

# Equal weights
df_qc["score"] = (
    df_qc["tx_rank"] +
    df_qc["prot_rank"] +
    df_qc["gen_rank"]
) / 3

# Final ranking
ranked_luad_qc = (
    df_qc
    .sort_values("score", ascending=False)
    .reset_index(drop=True)
)

ranked_luad_qc["rank"] = ranked_luad_qc.index + 1

print("Genes before QC:", len(merged_genes))
print("Genes after QC:", len(ranked_luad_qc))

display(
    ranked_luad_qc[
        [
            "rank", "gene",
            "rna_log2fc", "rna_pval",
            "prot_log2fc", "prot_pval",
            "mutation_freq", "n_pairs",
            "concordant", "score"
        ]
    ].head(15)
)
```

```python
known_luad = [
    "TP53", "KRAS", "EGFR", "STK11",
    "KEAP1", "BRAF", "MET", "ERBB2",
    "ALK", "NF1"
]

display(
    ranked_luad_qc[
        ranked_luad_qc["gene"].isin(known_luad)
    ][
        ["rank", "gene", "rna_log2fc", "prot_log2fc",
         "mutation_freq", "n_pairs", "concordant", "score"]
    ].sort_values("rank")
)
```

```python
# Final top 15 target table

targets_luad = ranked_luad_qc[
    [
        "rank",
        "gene",
        "ensembl",
        "rna_log2fc",
        "rna_pval",
        "prot_log2fc",
        "prot_pval",
        "mutation_freq",
        "n_pairs",
        "concordant",
        "score"
    ]
].head(15).copy()

output_file = Path("targets_luad.csv")

targets_luad.to_csv(output_file, index=False)

print("Saved:", output_file.resolve())
display(targets_luad)
```
